## 1. Unit of analysis + time window

*One row = one what, over which dates? State it, then verify it below.*

One row = one content webpage performance record for a specific date.

The dataset represents daily observations of content pages, including search performance, analytics availability, and engagement signals.

The analysis uses March 2026 as the time window because it is a mid-panel month. This avoids using the final month as development data and provides a realistic historical decision window.

## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*

### Features

The following fields will be used as model inputs:

- gsc_impressions: represents search visibility available before a refresh decision.
- gsc_clicks: represents historical clicks from search performance.
- gsc_sum_position: represents average ranking position.
- sessions: represents user traffic performance.
- scroll_events: represents user engagement behavior.

### Label

Content Refresh Opportunity Score.

This is a proxy label created from observed performance signals to estimate which pages have higher potential value from refreshing.

### Context

- content_hash_id: identifies the content page.
- client_has_gsc: indicates whether Search Console data exists.
- client_has_ga4: indicates whether analytics data exists.
- month: identifies the observation period.

### Excluded

- Future performance signals are excluded because they would not be available at the decision moment.
- Label-derived columns are excluded because they create data leakage and unrealistic model performance.
- client_hash_id is excluded because client identity is not relevant for the prediction task.

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*


This dataset supports decision-making about content refresh opportunities, but it cannot guarantee that updating a page will improve performance.

Limitations include:

- Some rows have missing analytics data because not all clients have GA4 availability.
- Search performance data depends on Search Console availability.
- The dataset represents historical observations and cannot capture future changes in search behavior.
- Daily observations may contain overlapping patterns that should be considered during modeling.
